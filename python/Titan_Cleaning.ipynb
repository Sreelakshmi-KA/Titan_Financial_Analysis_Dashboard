```python
# Read Profit & Loss sheet properly
profit_loss = pd.read_excel("Titan Company.xlsx", sheet_name="Profit & Loss",header=2)

profit_loss.head(5)
```

```python
# Check column names
print(profit_loss.columns.tolist())
```

```python
# Removing unwanted columns #
profit_loss = profit_loss.drop(columns=["Trailing", "Best Case", "Worst Case"])

# Renaming Date columns to FY format #
new_columns = ["Narration"] + [f"FY{col.year}" for col in profit_loss.columns[1:]]
profit_loss.columns = new_columns

print(profit_loss.columns.tolist())
```

```python
balance_sheet = pd.read_excel("Titan Company.xlsx", sheet_name="Balance Sheet",header=2)

balance_sheet = balance_sheet.drop(columns=["Trailing", "Best Case", "Worst Case"], errors="ignore")
new_columns = ["Narration"] + [f'FY{col.year}' for col in balance_sheet.columns[1:] if hasattr(col, "year")]
balance_sheet.columns = new_columns[:len(balance_sheet.columns)]

print(balance_sheet.columns.tolist())
```

```python
# Check if Balance Sheet has actual data in FY2017
print(balance_sheet[['Narration', 'FY2017', 'FY2018', 'FY2019']].head(10))
```

```python
# Check all NaN rows in balance sheet
print(balance_sheet.isnull().all(axis=1).sum(), "fully empty rows found")

# See which rows are NaN
print(balance_sheet)
```

```python
# Remove rows where ALL values are NaN
balance_sheet = balance_sheet.dropna(how='all')

# Check how many rows remain
print("Rows remaining:", len(balance_sheet))
print()
print(balance_sheet)
```

```python
# Check for any remaining NaN values
print("Remaining NaN values:")
print(balance_sheet.isnull().sum())
print()
print("Shape:", balance_sheet.shape)
print()
print(balance_sheet['Narration'].tolist())
```

```python
# Find which row has NaN in FY2017
print(balance_sheet[balance_sheet['FY2017'].isnull()])
```

```python
# Check which index positions have 'Total'
print(balance_sheet[balance_sheet['Narration'] == 'Total'])
```

```python
# Renaming the Total Rows #
balance_sheet.loc[4, 'Narration'] = 'Total_Liabilities_And_Equity'
balance_sheet.loc[10, 'Narration']= "Total_Assets"

print(balance_sheet['Narration'].tolist())
```

```python
# Reload Cash Flow with correct header3
cash_flow = pd.read_excel("Titan Company.xlsx", sheet_name="Cash Flow", header=2)

# Remove empty rows #
cash_flow = cash_flow.dropna(how='all')

# Remove Trailing/Best/Worst if they exist #
cash_flow = cash_flow.drop(columns=['Trailing', 'Best Case', 'Worst Case'], errors='ignore')

# Fix column names to FY format #
new_columns = ['Narration'] + [f'FY{col.year}' for col in cash_flow.columns[1:] if hasattr(col, 'year')]
cash_flow.columns = new_columns[:len(cash_flow.columns)]

# Verify #
print("Cash Flow columns:", cash_flow.columns.tolist())
print()
print("Cash Flow Narration:", cash_flow['Narration'].tolist())
```

```python
# Save all cleaned sheets to a new Excel file

with pd.ExcelWriter("Titan_Cleaned.xlsx") as writer:
    profit_loss.to_excel(writer, sheet_name="Profit_Loss", index=False)
    balance_sheet.to_excel(writer, sheet_name="Balance_Sheet", index=False)
    cash_flow.to_excel(writer, sheet_name="Cash_Flow", index=False)

print("Cleaned data saved successfully!! ✅")
```

```python
# ── FINAL PYTHON BLOCK ──────────────────────────

# Step 1 — Fix P&L numeric columns
year_cols = ['Revenue', 'Expenses', 'Operating_Profit', 'Other_Income', 
             'Depreciation', 'Interest', 'PBT', 'Net_Profit', 'EPS', 'EBITDA']
titan_pl[year_cols] = titan_pl[year_cols].apply(pd.to_numeric, errors='coerce')

# Step 2 — Calculate columns
titan_pl['Net_Profit_Margin'] = (titan_pl['Net_Profit'] / titan_pl['Revenue'] * 100).round(2)
titan_pl['YoY_Revenue_Growth'] = (titan_pl['Revenue'].pct_change() * 100).round(2)

# Step 3 — Fix Balance Sheet numeric columns
bs_cols = balance_sheet.columns[1:].tolist()
balance_sheet[bs_cols] = balance_sheet[bs_cols].apply(pd.to_numeric, errors='coerce')

# Step 4 — Save everything
with pd.ExcelWriter("Titan_Final.xlsx") as writer:
    titan_pl.to_excel(writer, sheet_name="Profit_Loss", index=False)
    balance_sheet.to_excel(writer, sheet_name="Balance_Sheet", index=False)
    cash_flow.to_excel(writer, sheet_name="Cash_Flow", index=False)

print("PYTHON PHASE COMPLETE!! 🎉")
print("Titan_Final.xlsx saved successfully!!")
print()
print(titan_pl[['Year', 'Revenue', 'Net_Profit', 'Net_Profit_Margin', 'YoY_Revenue_Growth']])
```
